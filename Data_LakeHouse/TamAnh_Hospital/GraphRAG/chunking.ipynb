```python
import pandas as pd

# Đường dẫn file CSV
file_path = "../Data/data_cleaned.csv"

# Đọc file
df = pd.read_csv(file_path)

# Chọn cột cần thống kê
col = "content"

# Tính số từ mỗi dòng
word_counts = df[col].fillna("").astype(str).str.split().str.len()

# In kết quả
print("Số dòng:", len(df))
print("Số từ tối thiểu:", int(word_counts.min()))
print("Số từ tối đa:", int(word_counts.max()))
print("Số từ trung bình:", round(word_counts.mean(), 2))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Đọc dữ liệu
df = pd.read_csv("../Data/data_cleaned.csv")

# Chọn cột content
splitter = RecursiveCharacterTextSplitter(
    chunk_size=1024,
    chunk_overlap=256,
    length_function=len,
    separators=["\n\n", "\n", " ", ""]
)

chunks = []
j = 0
for i, row in df.iterrows():
    text = str(row["content"]) if pd.notna(row["content"]) else ""
    if not text.strip():
        continue

    doc_chunks = splitter.split_text(text)
    for chunk in doc_chunks:
        chunks.append({
            "section_id": i,
            "chunk_id": j,
            "url": row.get("url", ""),
            "title": row.get("title", ""),
            "heading": row.get("heading", ""),
            "chunk_text": chunk
        })
        j += 1

chunk_df = pd.DataFrame(chunks)

print("Số chunk:", len(chunk_df))
chunk_df.head()
```

```python
chunk_df.to_csv("../Data/content_chunks.csv", index=False, encoding="utf-8-sig")
```
